# Perbandingan

```python
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.feature_selection import mutual_info_regression
import warnings
warnings.filterwarnings("ignore")

try:
    import dcor
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "dcor", "-q"])
    import dcor


# ------------------------------------------------------------------ parameter
alpha         = 0.05
lag_max       = 5
roll_window   = 30
n_perm_main   = 999    # permutasi untuk MI dan dCor
n_perm_lag    = 499    # permutasi per lag
n_mc_rolling  = 3000   # replikasi tingkat nol rolling dCor
SEED          = 42

rng = np.random.default_rng(SEED)


# ----------------------------------------------------------------- muat data
df = dfsentimen_final.copy()
df['tanggal'] = pd.to_datetime(df['tanggal'])
df = df.sort_values('tanggal').reset_index(drop=True)

# Kolom return dinamai 'return_ihsg' pada versi lama padahal dihitung dari
# harga penutupan ADRO. Baris berikut menerima kedua nama.
kol_ret = 'return_adro' if 'return_adro' in df.columns else 'return_ihsg'
df = df[['tanggal', 'close', kol_ret, 'vader', 'finbert']].dropna()

y       = df[kol_ret].values
vader   = df['vader'].values
finbert = df['finbert'].values
dates   = df['tanggal'].values
n_obs   = len(y)

print(f"n observasi : {n_obs:,}   "
      f"({df['tanggal'].min().date()} s.d. {df['tanggal'].max().date()})")
if kol_ret == 'return_ihsg':
    print("Catatan: kolom bernama 'return_ihsg' tetapi berisi imbal hasil ADRO. "
          "Sebaiknya diganti 'return_adro' agar tidak menyesatkan penguji.")


# --------------------------------------------------------------- fungsi bantu
def _mi(x, target):
    return float(mutual_info_regression(x.reshape(-1, 1), target,
                                        random_state=SEED)[0])


def _p_perm(obs, null):
    """p = (1 + #{null >= obs}) / (1 + B) — Phipson & Smyth (2010).
    Tidak pernah menghasilkan p = 0, berbeda dari rumus k/B."""
    null = np.asarray(null, dtype=float)
    return (1.0 + int((null >= obs - 1e-15).sum())) / (1.0 + null.size)
```

```python
# ==============================================================================
#  UJI 1 — MUTUAL INFORMATION
# ==============================================================================
print("\n" + "=" * 68)
print(" UJI 1 — MUTUAL INFORMATION")
print(" Ref: Kraskov, Stogbauer & Grassberger (2004), Phys. Rev. E 69, 066138")
print("=" * 68)

mi_vader   = _mi(vader,   y)
mi_finbert = _mi(finbert, y)

H_y         = _mi(y, y)                     # proksi entropi, bukan entropi asli
nmi_vader   = mi_vader   / H_y if H_y > 0 else 0
nmi_finbert = mi_finbert / H_y if H_y > 0 else 0

print(f"\n  Permutation test {n_perm_main}x ...")
_nv, _nf = [], []
for _ in range(n_perm_main):
    yp = rng.permutation(y)
    _nv.append(_mi(vader,   yp))
    _nf.append(_mi(finbert, yp))
pval_mi_v = _p_perm(mi_vader,   _nv)
pval_mi_f = _p_perm(mi_finbert, _nf)

print(f"\n  {'Model':<12} {'MI':>12} {'NMI':>12} {'p-value':>10} {'Signifikan':>12}")
print(f"  {'-'*62}")
print(f"  {'VADER':<12} {mi_vader:>12.6f} {nmi_vader:>12.6f} {pval_mi_v:>10.4f} "
      f"{('Ya *' if pval_mi_v < alpha else 'Tidak'):>12}")
print(f"  {'FinBERT':<12} {mi_finbert:>12.6f} {nmi_finbert:>12.6f} {pval_mi_f:>10.4f} "
      f"{('Ya *' if pval_mi_f < alpha else 'Tidak'):>12}")
print("\n  NMI dinormalisasi terhadap MI(y,y) sebagai proksi entropi. Bersifat")
print("  indikatif; peringkat antarmodel dibaca dari MI, bukan dari NMI.")
```

```python
# ==============================================================================
#  UJI 2 — DISTANCE CORRELATION
# ==============================================================================
print("\n" + "=" * 68)
print(" UJI 2 — DISTANCE CORRELATION (dCor)")
print(" Ref: Szekely, Rizzo & Bakirov (2007), Ann. Statist. 35(6), 2769-2794")
print("=" * 68)

dcor_vader   = dcor.distance_correlation(vader,   y)
dcor_finbert = dcor.distance_correlation(finbert, y)

print(f"\n  Permutation test {n_perm_main}x ...")
pval_dcor_v = dcor.independence.distance_covariance_test(
    vader, y, num_resamples=n_perm_main, random_state=SEED).pvalue
pval_dcor_f = dcor.independence.distance_covariance_test(
    finbert, y, num_resamples=n_perm_main, random_state=SEED).pvalue

print(f"\n  {'Model':<12} {'dCor':>10} {'p-value':>10} {'Signifikan':>12}")
print(f"  {'-'*48}")
print(f"  {'VADER':<12} {dcor_vader:>10.6f} {pval_dcor_v:>10.4f} "
      f"{('Ya *' if pval_dcor_v < alpha else 'Tidak'):>12}")
print(f"  {'FinBERT':<12} {dcor_finbert:>10.6f} {pval_dcor_f:>10.4f} "
      f"{('Ya *' if pval_dcor_f < alpha else 'Tidak'):>12}")
print("\n  dCor bernilai nol jika dan hanya jika kedua variabel benar-benar")
print("  independen — sifat yang tidak dimiliki korelasi Pearson. Uji ini")
print("  karenanya diberi bobot interpretatif terbesar.")
```

```python
# ==============================================================================
#  UJI 3 — SPEARMAN RANK CORRELATION
# ==============================================================================
print("\n" + "=" * 68)
print(" UJI 3 — SPEARMAN RANK CORRELATION (pembanding monoton)")
print("=" * 68)

rho_v, p_spv = stats.spearmanr(vader,   y)      # dua arah (default scipy)
rho_f, p_spf = stats.spearmanr(finbert, y)

print(f"\n  {'Model':<12} {'rho':>10} {'|rho|':>10} {'p-value':>10} {'Signifikan':>12}")
print(f"  {'-'*58}")
print(f"  {'VADER':<12} {rho_v:>10.4f} {abs(rho_v):>10.4f} {p_spv:>10.4f} "
      f"{('Ya *' if p_spv < alpha else 'Tidak'):>12}")
print(f"  {'FinBERT':<12} {rho_f:>10.4f} {abs(rho_f):>10.4f} {p_spf:>10.4f} "
      f"{('Ya *' if p_spf < alpha else 'Tidak'):>12}")
print("\n  Spearman hanya mendeteksi hubungan monoton. Koefisien mendekati nol")
print("  yang berdampingan dengan dCor signifikan justru menandakan dependensi")
print("  bersifat non-linear sekaligus non-monoton.")
```

```python
# ==============================================================================
#  UJI 4 — DIRECTIONAL ACCURACY
# ==============================================================================
print("\n" + "=" * 68)
print(" UJI 4 — DIRECTIONAL ACCURACY + UJI BINOMIAL DUA ARAH")
print("=" * 68)

prop_pos    = float((y > 0).mean())
base_mayor  = max(prop_pos, 1 - prop_pos)


def _da(skor, target):
    """Kesesuaian tanda. Binomial DUA ARAH agar penyimpangan ke bawah 0,5
    juga terdeteksi. Versi lama memakai 'greater' sehingga tidak bisa."""
    k = int((np.sign(skor) == np.sign(target)).sum())   # hitung langsung,
    n = len(target)                                     # hindari int(da*n)
    return k / n, stats.binomtest(k, n, 0.5, alternative='two-sided').pvalue


dir_v,  pval_da_v  = _da(vader,   y)
dir_f,  pval_da_f  = _da(finbert, y)
dir_vc, pval_da_vc = _da(vader   - vader.mean(),   y)   # skor terpusat
dir_fc, pval_da_fc = _da(finbert - finbert.mean(), y)

print(f"\n  {'Spesifikasi':<26} {'Dir.Acc':>9} {'vs 0,5':>10} {'p-value':>11} {'Sig.':>8}")
print(f"  {'-'*68}")
for nama, d_, p_ in [("VADER", dir_v, pval_da_v),
                     ("FinBERT", dir_f, pval_da_f),
                     ("VADER (skor terpusat)", dir_vc, pval_da_vc),
                     ("FinBERT (skor terpusat)", dir_fc, pval_da_fc)]:
    pt = "< 0.0001" if p_ < 1e-4 else f"{p_:.4f}"
    print(f"  {nama:<26} {d_:>9.4f} {d_-0.5:>+10.4f} {pt:>11} "
          f"{('Ya *' if p_ < alpha else 'Tidak'):>8}")
print(f"  {'Baseline acak':<26} {0.5:>9.4f}")
print(f"  {'Baseline mayoritas arah':<26} {base_mayor:>9.4f} {base_mayor-0.5:>+10.4f}")

print(f"\n  Proporsi hari imbal hasil positif : {prop_pos:.4f}")
if abs(dir_v - prop_pos) < 0.01 or abs(dir_f - prop_pos) < 0.01:
    print("  PERHATIAN: DA nyaris sama dengan proporsi di atas. Ketika skor")
    print("  sentimen hampir selalu bertanda tetap, yang terukur adalah")
    print("  komposisi tanda sampel, bukan daya diskriminasi model.")
print(f"  Ambang kompetitif menurut Atsalakis & Valavanis (2009) : 0.5500")
```

```python
# ==============================================================================
#  UJI 5 — LAG-BASED MUTUAL INFORMATION
# ==============================================================================
print("\n" + "=" * 68)
print(" UJI 5 — LAG-BASED MUTUAL INFORMATION (indikator pendahulu)")
print("=" * 68)
print(f"\n  Permutation test {n_perm_lag}x per lag ...")

lag_mi_v, lag_mi_f = {}, {}
lag_p_v,  lag_p_f  = {}, {}
bonferroni = alpha / (lag_max + 1)

for lag in range(0, lag_max + 1):
    yy = y[lag:]        if lag > 0 else y
    vv = vader[:-lag]   if lag > 0 else vader
    ff = finbert[:-lag] if lag > 0 else finbert

    m_v, m_f = _mi(vv, yy), _mi(ff, yy)
    nv, nf = [], []
    for _ in range(n_perm_lag):
        yp = rng.permutation(yy)
        nv.append(_mi(vv, yp))
        nf.append(_mi(ff, yp))

    lag_mi_v[lag], lag_mi_f[lag] = m_v, m_f
    lag_p_v[lag],  lag_p_f[lag]  = _p_perm(m_v, nv), _p_perm(m_f, nf)

print(f"\n  {'Lag':<5} {'MI VADER':>12} {'p VADER':>10} {'MI FinBERT':>13} "
      f"{'p FinBERT':>11} {'Unggul':>10}")
print(f"  {'-'*68}")
for lag in range(0, lag_max + 1):
    mv, mf = lag_mi_v[lag], lag_mi_f[lag]
    ung = "VADER" if mv > mf else ("FinBERT" if mf > mv else "Seimbang")
    print(f"  {lag:<5} {mv:>12.6f} {lag_p_v[lag]:>10.4f} {mf:>13.6f} "
          f"{lag_p_f[lag]:>11.4f} {ung:>10}")

sig_lag_v = [l for l in lag_mi_v if lag_p_v[l] < alpha]
sig_lag_f = [l for l in lag_mi_f if lag_p_f[l] < alpha]
best_v = max(sig_lag_v, key=lambda l: lag_mi_v[l]) if sig_lag_v else max(lag_mi_v, key=lag_mi_v.get)
best_f = max(sig_lag_f, key=lambda l: lag_mi_f[l]) if sig_lag_f else max(lag_mi_f, key=lag_mi_f.get)

print(f"\n  Ambang Bonferroni ({alpha} / {lag_max + 1} lag) : {bonferroni:.4f}")
print(f"  VADER   signifikan nominal pada lag {sig_lag_v}; "
      f"bertahan Bonferroni pada lag {[l for l in sig_lag_v if lag_p_v[l] < bonferroni]}")
print(f"  FinBERT signifikan nominal pada lag {sig_lag_f}; "
      f"bertahan Bonferroni pada lag {[l for l in sig_lag_f if lag_p_f[l] < bonferroni]}")
print(f"\n  VADER   MI terkuat: lag {best_v} (MI={lag_mi_v[best_v]:.6f}, p={lag_p_v[best_v]:.4f})")
print(f"  FinBERT MI terkuat: lag {best_f} (MI={lag_mi_f[best_f]:.6f}, p={lag_p_f[best_f]:.4f})")
print(f"\n  Nilai p pada lag 0 dapat berbeda tipis dari Uji 1 karena memakai")
print(f"  {n_perm_lag} replikasi, sedangkan Uji 1 memakai {n_perm_main}. Nilai MI-nya identik.")
print("\n  MI bernilai 0,000000 adalah konsekuensi estimator k-tetangga terdekat")
print("  yang memangkas estimasi negatif menjadi nol, bukan bukti independensi.")
```

```python
# ==============================================================================
#  UJI 6 — ROLLING DISTANCE CORRELATION
# ==============================================================================
print("\n" + "=" * 68)
print(f" UJI 6 — ROLLING DISTANCE CORRELATION (jendela {roll_window} hari)")
print("=" * 68)

roll_dcor_v, roll_dcor_f, roll_dates = [], [], []
for i in range(roll_window, n_obs + 1):
    sl = slice(i - roll_window, i)
    roll_dcor_v.append(dcor.distance_correlation(vader[sl],   y[sl]))
    roll_dcor_f.append(dcor.distance_correlation(finbert[sl], y[sl]))
    roll_dates.append(dates[i - 1])

roll_dcor_v = np.array(roll_dcor_v)
roll_dcor_f = np.array(roll_dcor_f)

# Tingkat nol: dCor pada n kecil bias ke atas, jadi rata-rata 0,3x sama sekali
# tidak berarti "dependensi kuat". Tanpa pembanding ini, Uji 6 tidak terbaca.
print(f"\n  Simulasi tingkat nol {n_mc_rolling}x ...")
null_roll = np.array([
    dcor.distance_correlation(rng.standard_normal(roll_window),
                              rng.standard_normal(roll_window))
    for _ in range(n_mc_rolling)
])
null_mean = float(null_roll.mean())
null_p95  = float(np.percentile(null_roll, 95))

print(f"\n  {'Deret':<38} {'Mean':>10} {'Persentil 95':>14}")
print(f"  {'-'*64}")
print(f"  {'VADER (empiris)':<38} {roll_dcor_v.mean():>10.4f} {'—':>14}")
print(f"  {'FinBERT (empiris)':<38} {roll_dcor_f.mean():>10.4f} {'—':>14}")
print(f"  {'Tingkat nol (deret independen, n=30)':<38} {null_mean:>10.4f} {null_p95:>14.4f}")

roll_menentukan = (roll_dcor_v.mean() > null_p95) or (roll_dcor_f.mean() > null_p95)
print(f"\n  Jumlah jendela : {len(roll_dcor_v):,}")
print(f"  Selisih VADER - FinBERT : {roll_dcor_v.mean() - roll_dcor_f.mean():+.4f}")
if not roll_menentukan:
    print("\n  KESIMPULAN: kedua nilai empiris berada di dalam sebaran tingkat nol,")
    print("  sehingga uji ini TIDAK INFORMATIF untuk membedakan kedua model pada")
    print("  spesifikasi jendela 30 hari. Selisih di antara keduanya tidak dapat")
    print("  dijadikan dasar kesimpulan apa pun. Hasil tetap dilaporkan apa adanya.")
    print("  Saran lanjutan: estimator dCor terkoreksi bias, atau jendela 120 hari.")
else:
    print("\n  Sekurang-kurangnya satu nilai empiris melampaui persentil ke-95")
    print("  tingkat nol, sehingga uji ini informatif.")
```

```python
# ==============================================================================
#  REKAPITULASI DAN PENETAPAN MODEL
# ==============================================================================
print("\n" + "=" * 68)
print(" REKAPITULASI — SCORECARD ENAM UJI")
print("=" * 68)

# Satu uji = satu suara. Versi lama memecah NMI dan nilai p menjadi baris
# tersendiri, sehingga Mutual Information praktis dihitung dua kali dan
# Distance Correlation tiga kali.
metrics = [
    ("Mutual Information",     mi_vader,             mi_finbert,             pval_mi_v,  pval_mi_f,  True),
    ("Distance Correlation",   dcor_vader,           dcor_finbert,           pval_dcor_v, pval_dcor_f, True),
    ("Spearman |rho|",         abs(rho_v),           abs(rho_f),             p_spv,      p_spf,      True),
    ("Directional Accuracy",   dir_v,                dir_f,                  pval_da_v,  pval_da_f,  True),
    ("Lag-based MI (puncak)",  lag_mi_v[best_v],     lag_mi_f[best_f],       lag_p_v[best_v], lag_p_f[best_f], True),
    ("Rolling dCor (30 hari)", roll_dcor_v.mean(),   roll_dcor_f.mean(),     np.nan,     np.nan,     roll_menentukan),
]

skor_v, skor_f = 0, 0
print(f"\n  {'Uji':<24} {'VADER':>20} {'FinBERT':>20} {'Unggul':>18}")
print(f"  {'-'*84}")
for nama, vv, ff, pv, pf, dihitung in metrics:
    if not dihitung:
        unggul = "Tidak menentukan"
        sv, sf = f"{vv:.6f}", f"{ff:.6f}"
    else:
        unggul = "VADER" if vv > ff else "FinBERT"
        skor_v += (unggul == "VADER")
        skor_f += (unggul == "FinBERT")
        _f = lambda p: ("< .0001" if p < 1e-4 else f"{p:.4f}")
        sv = f"{vv:.6f} (p={_f(pv)})"
        sf = f"{ff:.6f} (p={_f(pf)})"
    print(f"  {nama:<24} {sv:>20} {sf:>20} {unggul:>18}")
print(f"  {'-'*84}")
print(f"  {'Skor akhir':<24} {skor_v:>20} {skor_f:>20}")

# ---------------------------------------------------- aturan keputusan a priori
print("\n" + "=" * 68)
print(" ATURAN KEPUTUSAN BERJENJANG (ditetapkan sebelum hasil diketahui)")
print("=" * 68)

formal = [("Mutual Information", pval_mi_v, pval_mi_f),
          ("Distance Correlation", pval_dcor_v, pval_dcor_f),
          ("Lag-based MI", lag_p_v[best_v], lag_p_f[best_f])]
sig_v_formal = [n for n, pv, _ in formal if pv < alpha]
sig_f_formal = [n for n, _, pf in formal if pf < alpha]

print(f"\n  Kriteria 1 — signifikansi pada uji berhipotesis formal:")
print(f"     VADER   : {len(sig_v_formal)}/3  {sig_v_formal}")
print(f"     FinBERT : {len(sig_f_formal)}/3  {sig_f_formal}")
print(f"\n  Kriteria 2 — bobot interpretatif dCor:")
print(f"     VADER   dCor = {dcor_vader:.6f} (p={pval_dcor_v:.4f})")
print(f"     FinBERT dCor = {dcor_finbert:.6f} (p={pval_dcor_f:.4f})")
print(f"\n  Kriteria 3 — konsistensi lintas horizon waktu:")
print(f"     VADER   signifikan pada lag {sig_lag_v}")
print(f"     FinBERT signifikan pada lag {sig_lag_f}")

if len(sig_v_formal) != len(sig_f_formal):
    pemenang = "VADER" if len(sig_v_formal) > len(sig_f_formal) else "FinBERT"
    dasar = "jumlah uji berhipotesis formal yang signifikan"
else:
    pemenang = "VADER" if dcor_vader > dcor_finbert else "FinBERT"
    dasar = "nilai dCor, karena kriteria 1 imbang"

print(f"\n  >>> MODEL TERPILIH: {pemenang}")
print(f"      Dasar : {dasar}.")
print(f"      Keunggulan nilai tanpa dukungan signifikansi tidak dijadikan")
print(f"      dasar seleksi fitur.")

tanpa_da = sum(vv > ff for nama, vv, ff, _, _, d in metrics
               if d and nama != "Directional Accuracy")
n_dihitung = sum(1 for *_, d in metrics if d) - 1
print(f"\n  Robustness: tanpa Directional Accuracy, skor menjadi "
      f"{tanpa_da} berbanding {n_dihitung - tanpa_da}"
      f"{' dengan rolling dCor tidak menentukan' if not roll_menentukan else ''}.")

# Variabel kompatibilitas untuk sel visualisasi
n_uji_dihitung = sum(1 for *_, d in metrics if d)
print("\n Selesai.")
```

```python
# ==============================================================================
# ==============================================================================
#  SEL "VISUALISASI" — PENYESUAIAN MINIMAL
#  Hanya empat panel yang berubah. Sisanya identik dengan versi lama.
#    Panel 1 & 2 : label nilai p ditambahkan
#    Panel 3     : baseline mayoritas arah ditambahkan
#    Panel 4     : lag signifikan ditandai
#    Panel 5     : scorecard /6, bukan /8
#    Panel 6     : garis tingkat nol dan pita persentil ke-95 ditambahkan
# ==============================================================================

import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

COLOR_V = "#4fc3f7"
COLOR_F = "#f48fb1"
BG      = "#161b22"
TEXT    = "#e6edf3"
GRID    = "#30363d"
NULLC   = "#ffb74d"


def style_ax(ax, title):
    ax.set_facecolor(BG)
    ax.set_title(title, color=TEXT, fontsize=9, fontweight='bold', pad=7)
    ax.tick_params(colors=TEXT, labelsize=8)
    ax.xaxis.label.set_color(TEXT)
    ax.yaxis.label.set_color(TEXT)
    for sp in ax.spines.values():
        sp.set_edgecolor(GRID)
    ax.grid(True, color=GRID, linewidth=0.5, alpha=0.6)


fig = plt.figure(figsize=(18, 13))
fig.patch.set_facecolor('#0d1117')
gs = gridspec.GridSpec(3, 3, figure=fig, hspace=0.48, wspace=0.35)

# Panel 1 — MI, kini dengan nilai p
ax1 = fig.add_subplot(gs[0, 0])
bars = ax1.bar(["VADER", "FinBERT"], [mi_vader, mi_finbert],
               color=[COLOR_V, COLOR_F], width=0.5, edgecolor=GRID)
for b, v, p in zip(bars, [mi_vader, mi_finbert], [pval_mi_v, pval_mi_f]):
    ax1.text(b.get_x() + b.get_width()/2, b.get_height() + 1e-5,
             f"{v:.5f}\np={p:.4f}", ha='center', va='bottom',
             color=(TEXT if p >= alpha else "#7ee787"), fontsize=8)
style_ax(ax1, "Mutual Information\n(↑ lebih baik; p dari permutation test)")
ax1.set_ylabel("MI Score", color=TEXT)
ax1.set_ylim(0, max(mi_vader, mi_finbert) * 1.45 + 1e-6)

# Panel 2 — dCor, kini dengan nilai p
ax2 = fig.add_subplot(gs[0, 1])
bars2 = ax2.bar(["VADER", "FinBERT"], [dcor_vader, dcor_finbert],
                color=[COLOR_V, COLOR_F], width=0.5, edgecolor=GRID)
for b, v, p in zip(bars2, [dcor_vader, dcor_finbert], [pval_dcor_v, pval_dcor_f]):
    ax2.text(b.get_x() + b.get_width()/2, b.get_height() + 5e-4,
             f"{v:.4f}\np={p:.4f}", ha='center', va='bottom',
             color=(TEXT if p >= alpha else "#7ee787"), fontsize=8)
style_ax(ax2, "Distance Correlation (dCor)\n(↑ lebih baik; nol ⟺ independen)")
ax2.set_ylabel("dCor", color=TEXT)
ax2.set_ylim(0, max(dcor_vader, dcor_finbert) * 1.45)

# Panel 3 — DA dengan dua baseline
ax3 = fig.add_subplot(gs[0, 2])
bars3 = ax3.bar(["VADER", "FinBERT"], [dir_v, dir_f],
                color=[COLOR_V, COLOR_F], width=0.5, edgecolor=GRID)
for b, v in zip(bars3, [dir_v, dir_f]):
    ax3.text(b.get_x() + b.get_width()/2, b.get_height() + 0.004,
             f"{v:.4f}", ha='center', va='bottom', color=TEXT, fontsize=8)
ax3.axhline(0.5, color=NULLC, lw=1.2, ls='--', label="Tebakan acak (0,500)")
ax3.axhline(base_mayor, color="#f85149", lw=1.2, ls=':',
            label=f"Mayoritas arah ({base_mayor:.3f})")
ax3.legend(fontsize=7, labelcolor=TEXT, facecolor=BG, edgecolor=GRID, loc='upper right')
style_ax(ax3, "Directional Accuracy\n(dibandingkan DUA baseline)")
ax3.set_ylabel("Accuracy", color=TEXT)
ax3.set_ylim(0.35, 0.70)

# Panel 4 — Lag MI, lag signifikan ditandai
ax4 = fig.add_subplot(gs[1, :2])
lags = list(lag_mi_v.keys())
x = np.arange(len(lags)); w = 0.35
b_v = ax4.bar(x - w/2, [lag_mi_v[l] for l in lags], w, label='VADER',
              color=COLOR_V, alpha=0.85, edgecolor=GRID)
b_f = ax4.bar(x + w/2, [lag_mi_f[l] for l in lags], w, label='FinBERT',
              color=COLOR_F, alpha=0.85, edgecolor=GRID)
for l, b in zip(lags, b_v):
    if lag_p_v[l] < alpha:
        ax4.text(b.get_x() + b.get_width()/2, b.get_height(), "*",
                 ha='center', va='bottom', color="#7ee787", fontsize=14)
for l, b in zip(lags, b_f):
    if lag_p_f[l] < alpha:
        ax4.text(b.get_x() + b.get_width()/2, b.get_height(), "*",
                 ha='center', va='bottom', color="#7ee787", fontsize=14)
ax4.set_xticks(x)
ax4.set_xticklabels([f"Lag {l}" for l in lags])
ax4.legend(fontsize=8, labelcolor=TEXT, facecolor=BG, edgecolor=GRID)
style_ax(ax4, "Lag-based Mutual Information — Uji Indikator Pendahulu\n"
              "(tanda * = signifikan pada permutation test, α = 0,05)")
ax4.set_ylabel("MI Score", color=TEXT)

# Panel 5 — Scorecard enam uji
ax5 = fig.add_subplot(gs[1, 2])
ax5.set_facecolor(BG)
ax5.axis('off')
txt = (f"SCORECARD\n{'─'*26}\n"
       f"Dari 6 uji, {n_uji_dihitung} menentukan\n"
       f"VADER    : {skor_v}/{n_uji_dihitung}\n"
       f"FinBERT  : {skor_f}/{n_uji_dihitung}\n"
       f"{'─'*26}\n\n"
       f"Uji berhipotesis formal:\n"
       f"  VADER   {len(sig_v_formal)}/3\n"
       f"  FinBERT {len(sig_f_formal)}/3\n\n"
       f"MODEL TERPILIH:\n  {pemenang}\n\n"
       f"Dasar:\n{dasar}\n\n"
       f"Rolling dCor:\n"
       f"  {'tidak menentukan' if not roll_menentukan else 'menentukan'}")
ax5.text(0.05, 0.95, txt, transform=ax5.transAxes, fontsize=8,
         va='top', fontfamily='monospace', color=TEXT,
         bbox=dict(boxstyle='round', facecolor='#21262d',
                   edgecolor='#388bfd', alpha=0.9))
ax5.set_title("Ringkasan", color=TEXT, fontsize=9, fontweight='bold', pad=7)

# Panel 6 — Rolling dCor dengan tingkat nol
ax6 = fig.add_subplot(gs[2, :])
rd = pd.to_datetime(roll_dates)
ax6.plot(rd, roll_dcor_v, color=COLOR_V, lw=1.0,
         label=f"VADER (mean={roll_dcor_v.mean():.4f})")
ax6.plot(rd, roll_dcor_f, color=COLOR_F, lw=1.0,
         label=f"FinBERT (mean={roll_dcor_f.mean():.4f})")
ax6.axhline(null_mean, color=NULLC, lw=1.6, ls='--',
            label=f"Rata-rata tingkat nol ({null_mean:.4f})")
ax6.axhline(null_p95, color="#f85149", lw=1.2, ls=':',
            label=f"Persentil ke-95 tingkat nol ({null_p95:.4f})")
ax6.axhspan(0, null_p95, color=NULLC, alpha=0.07)
ax6.legend(fontsize=8, labelcolor=TEXT, facecolor=BG, edgecolor=GRID,
           loc='upper right', ncol=2)
style_ax(ax6, f"Rolling Distance Correlation (jendela {roll_window} hari)\n"
              "Area berarsir = sebaran tingkat nol; nilai di dalamnya "
              "tidak dapat dibedakan dari kebetulan")
ax6.set_ylabel("dCor", color=TEXT)
ax6.set_xlabel("Tanggal", color=TEXT)

fig.suptitle(
    "Perbandingan Non-Linear: VADER vs FinBERT terhadap Imbal Hasil ADRO\n"
    "Mutual Information · Distance Correlation · Directional Accuracy · "
    "Lag-based MI · Rolling dCor  |  "
    "Ref: Kraskov et al. (2004) · Székely et al. (2007) · Dunn (1961)",
    color=TEXT, fontsize=10, fontweight='bold', y=0.995
)

plt.savefig("sentiment_nonlinear_comparison.png", dpi=150,
            bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
print("\n Selesai.")
```

```python

from yahooquery import Ticker
import pandas as pd

ticker = "ADRO.JK"
t = Ticker(ticker)

# Tambah 1 tahun ke belakang (2015)
adro = t.history(start="2015-01-01", end="2025-12-31").reset_index()

adro = adro[["date", "symbol", "close", "volume", "high", "low"]]
adro.rename(columns={
    "date": "tanggal",
    "symbol": "ticker"
}, inplace=True)

adro["close"] = adro["close"].round(0).astype(int)
import numpy as np
adro["SMA_10"] = adro["close"].rolling(window=10).mean()

weights = np.arange(1, 11)
adro["WMA_10"] = adro["close"].rolling(10).apply(lambda x: np.dot(x, weights)/weights.sum(), raw=True)

ema12 = adro["close"].ewm(span=12, adjust=False).mean()
ema26 = adro["close"].ewm(span=26, adjust=False).mean()
adro["MACD"] = ema12 - ema26

adro["Momentum"] = adro["close"] - adro["close"].shift(10)

low14 = adro["low"].rolling(14).min()
high14 = adro["high"].rolling(14).max()

adro["Stoch_K"] = ((adro["close"] - low14) / (high14 - low14)) * 100

adro["Stoch_D"] = adro["Stoch_K"].rolling(3).mean()

delta = adro["close"].diff()
gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.rolling(14).mean()
avg_loss = loss.rolling(14).mean()

rs = avg_gain / avg_loss
adro["RSI"] = 100 - (100 / (1 + rs))

adro["Williams_R"] = ((high14 - adro["close"]) / (high14 - low14)) * -100

tp = (adro["high"] + adro["low"] + adro["close"]) / 3
sma_tp = tp.rolling(20).mean()
mad = tp.rolling(20).apply(lambda x: np.mean(np.abs(x - np.mean(x))))

adro["CCI"] = (tp - sma_tp) / (0.015 * mad)

mf = tp * adro["volume"]

positive_mf = mf.where(tp > tp.shift(1), 0)
negative_mf = mf.where(tp < tp.shift(1), 0)

pos_mf_sum = positive_mf.rolling(14).sum()
neg_mf_sum = negative_mf.rolling(14).sum()

mfr = pos_mf_sum / neg_mf_sum
adro["MFI"] = 100 - (100 / (1 + mfr))

adro["tanggal"] = pd.to_datetime(adro["tanggal"])

adro = adro[adro["tanggal"] >= "2016-01-01"]
indikator = adro[[
    "tanggal",
    "SMA_10", "WMA_10", "MACD", "Momentum",
    "Stoch_K", "Stoch_D", "RSI",
    "Williams_R", "CCI", "MFI"
]]
dfsentimen_final = pd.merge(
    dfsentimen_final,
    indikator,
    on="tanggal",
    how="left"
)
```
